```python
import os
# Prevent Windows Anaconda + PyTorch OpenMP (libiomp5md.dll) kernel crash:
os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Interactive Worksheet Inputs:
# Rate your current confidence in each domain from 1 (Brand New / Low Confidence) to 5 (Expert / High Confidence),
# and set your total available study hours before exam day.

TOTAL_STUDY_HOURS = 30.0  # <-- Customize your total study hours (e.g., 20, 30, 40)

USER_CONFIDENCE_RATINGS = {
    "Core Machine Learning & AI Knowledge": 2,  # 1 (Low) to 5 (High)
    "Software Development": 3,
    "Experimentation": 2,
    "Data Analysis & Visualization": 3,
    "Trustworthy AI": 4,
}

blueprint_data = [
    {
        "Domain": "Core Machine Learning & AI Knowledge",
        "Weight_Pct": 30.0,
        "Chapters": "Ch 2, Ch 8",
        "Key_Topics": "ML paradigms, Overfitting/Regularization, GPU vs CPU, Precision/VRAM, Transformers (BERT/GPT/T5), Tokenizers, Quantization, TensorRT-LLM",
        "Labs": "2-01, 2-02, 8-01, 8-02, 8-03",
    },
    {
        "Domain": "Software Development",
        "Weight_Pct": 24.0,
        "Chapters": "Ch 1, Ch 3",
        "Key_Topics": "Python idioms, HF Transformers & spaCy, LangChain/LlamaIndex, RAG pipelines, Vector DBs, Streaming (TTFT/ITL), Docker & NVIDIA NIM",
        "Labs": "1-01, 3-01, 3-02, 3-03, 3-04",
    },
    {
        "Domain": "Experimentation",
        "Weight_Pct": 22.0,
        "Chapters": "Ch 4, Ch 7",
        "Key_Topics": "Hypothesis design, Data splits/leakage, LoRA/QLoRA, BLEU/ROUGE/Perplexity, LM Eval Harness, Zero/Few-shot, CoT, Sampling, Prompt Injection",
        "Labs": "4-01, 4-02, 4-03, 7-01, 7-02",
    },
    {
        "Domain": "Data Analysis & Visualization",
        "Weight_Pct": 14.0,
        "Chapters": "Ch 5",
        "Key_Topics": "Text corpus EDA, Zipf's law, Context window percentiles, Deduplication, RAPIDS cudf.pandas, PCA vs t-SNE vs UMAP, Chart selection",
        "Labs": "5-01, 5-02",
    },
    {
        "Domain": "Trustworthy AI",
        "Weight_Pct": 10.0,
        "Chapters": "Ch 6",
        "Key_Topics": "Hallucinations, Bias & Fairness probes, Privacy/PII, NeMo Guardrails (Colang), Explainability (SHAP/LIME), Model Card++, NIST AI RMF",
        "Labs": "6-01, 6-02, 6-03",
    },
]

df_plan = pd.DataFrame(blueprint_data)
df_plan["Est_Questions"] = (df_plan["Weight_Pct"] / 100.0 * 50).round().astype(int)
df_plan["Confidence_1_to_5"] = df_plan["Domain"].map(USER_CONFIDENCE_RATINGS)

# Priority Gap Multiplier: (6 - Confidence) ranges from 5 (for Confidence=1) down to 1 (for Confidence=5)
df_plan["Knowledge_Gap_Factor"] = 6 - df_plan["Confidence_1_to_5"]
df_plan["Weighted_Priority_Score"] = df_plan["Weight_Pct"] * df_plan["Knowledge_Gap_Factor"]
df_plan["Recommended_Study_Pct"] = (
    df_plan["Weighted_Priority_Score"] / df_plan["Weighted_Priority_Score"].sum() * 100.0
).round(1)
df_plan["Recommended_Hours"] = (
    df_plan["Weighted_Priority_Score"] / df_plan["Weighted_Priority_Score"].sum() * TOTAL_STUDY_HOURS
).round(1)

display_cols = [
    "Domain", "Weight_Pct", "Est_Questions", "Chapters",
    "Confidence_1_to_5", "Recommended_Study_Pct", "Recommended_Hours", "Labs"
]
print(f"=== PERSONALIZED NCA-GENL STUDY PLAN (Total Budget: {TOTAL_STUDY_HOURS} Hours) ===")
print(df_plan[display_cols].to_string(index=False))
```

## 2. Visualizing Blueprint Weight vs. Personalized Study Allocation

In **Chapter 1 (*Using the Weights to Plan Study*)**, the book explains why studying every topic for an equal number of hours is a common pitfall:
- A low-confidence rating in a **30% domain** (Core ML & AI Knowledge) carries **3× the exam risk** of the same confidence gap in a **10% domain** (Trustworthy AI).
- By weighting each domain's blueprint percentage by your personal knowledge gap $(6 - 	ext{Confidence})$, your study plan automatically directs extra hours toward high-weight, low-confidence domains while maintaining review coverage for smaller domains.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

y_pos = np.arange(len(df_plan))
bar_height = 0.36

# Left chart: Official Blueprint Weight (%) vs Recommended Study Allocation (%)
axes[0].barh(y_pos + bar_height / 2, df_plan["Weight_Pct"], height=bar_height, label="Official Exam Weight (%)", color="#76B900")
axes[0].barh(y_pos - bar_height / 2, df_plan["Recommended_Study_Pct"], height=bar_height, label="Your Study Allocation (%)", color="#1A73E8")
axes[0].set_yticks(y_pos)
axes[0].set_yticklabels(df_plan["Domain"])
axes[0].invert_yaxis()
axes[0].set_xlabel("Percentage (%)")
axes[0].set_title("Official Blueprint Weight vs. Gap-Adjusted Study Share")
axes[0].legend(loc="lower right")
axes[0].grid(axis="x", linestyle="--", alpha=0.5)

# Right chart: Recommended Study Hours by Domain + Confidence Annotations
bars = axes[1].barh(y_pos, df_plan["Recommended_Hours"], color="#2E7D32", alpha=0.85)
axes[1].set_yticks(y_pos)
axes[1].set_yticklabels(df_plan["Chapters"] + " (" + df_plan["Domain"].str.split().str[0] + "...)")
axes[1].invert_yaxis()
axes[1].set_xlabel("Recommended Study Hours")
axes[1].set_title(f"Hours Allocation Across {TOTAL_STUDY_HOURS:.0f}-Hour Study Budget")
axes[1].grid(axis="x", linestyle="--", alpha=0.5)

for bar, conf, hrs in zip(bars, df_plan["Confidence_1_to_5"], df_plan["Recommended_Hours"]):
    axes[1].text(bar.get_width() + 0.2, bar.get_y() + bar.get_height() / 2,
                 f"{hrs:.1f}h (Conf: {conf}/5)", va="center", fontsize=9)

axes[1].set_xlim(0, max(df_plan["Recommended_Hours"]) * 1.35)
plt.tight_layout()
plt.show()
```

## 4. Simulating the Two-Pass Exam Pacing Strategy

In **Chapter 1 (*Study Strategies and Exam-Taking Best Practices*)**, the book recommends the **Two-Pass Method** for the 60-minute, 50-question NCA-GENL exam:
- **Pass 1 (Minutes 0–40):** Move steadily through all 50 questions (~45 seconds per question). Answer every question you know immediately. If a scenario question requires long calculation or hesitation, **select your best educated guess, flag the question for review, and move on immediately**.
- **Pass 2 (Minutes 40–55):** Return only to your flagged questions with full visibility of how much time remains on the clock.
- **Final Check (Minutes 55–60):** Confirm no question was left blank (there is **no penalty for guessing** on NCA-GENL).

Let's run a quick simulation comparing **Linear Pacing (getting stuck on hard questions early)** vs. the **Two-Pass Strategy**.

```python
def simulate_exam_pacing(seed: int = 42):
    rng = np.random.default_rng(seed)
    # 50 questions: 38 standard (~40s needed, 85% accuracy), 12 complex (~140s if stuck, 60% accuracy if given time, 25% random guess)
    is_complex = np.zeros(50, dtype=bool)
    is_complex[rng.choice(50, size=12, replace=False)] = True

    # Strategy A: Linear (never skips complex questions; runs out of time before reaching easy questions at the end)
    time_left_linear = 3600.0  # 60 minutes in seconds
    expected_correct_linear = 0.0
    answered_linear = 0
    for q_complex in is_complex:
        t_needed = 215.0 if q_complex else 52.0
        if time_left_linear >= t_needed:
            time_left_linear -= t_needed
            expected_correct_linear += (0.60 if q_complex else 0.85)
            answered_linear += 1
        else:
            # Forced to rush/guess remaining questions at the end!
            expected_correct_linear += 0.25

    # Strategy B: Two-Pass Method (answers all 38 standard questions in Pass 1 + educated guess on complex, then solves flagged in Pass 2)
    pass1_time = 38 * 42.0 + 12 * 25.0  # 42s on standard, 25s to read/guess/flag complex = 1896s (~31.6 min)
    remaining_for_pass2 = 3600.0 - pass1_time
    complex_fully_reviewed = min(12, int(remaining_for_pass2 // 95.0))
    complex_guessed_only = 12 - complex_fully_reviewed
    expected_correct_twopass = (38 * 0.85) + (complex_fully_reviewed * 0.60) + (complex_guessed_only * 0.35)

    print("=== 60-MINUTE, 50-QUESTION EXAM PACING COMPARISON ===")
    print(f"Strategy A (Linear / Getting Stuck Early):")
    print(f"  Questions carefully read before clock expired : {answered_linear}/50 (Rushed/Guessed last {50 - answered_linear} questions!)")
    print(f"  Expected Score                                : {expected_correct_linear / 50 * 100:.1f}% ({expected_correct_linear:.1f}/50)")
    print(f"\nStrategy B (Two-Pass Method from Chapter 1):")
    print(f"  Standard questions answered in Pass 1         : 38/38")
    print(f"  Flagged complex questions solved in Pass 2    : {complex_fully_reviewed}/12")
    print(f"  Expected Score                                : {expected_correct_twopass / 50 * 100:.1f}% ({expected_correct_twopass:.1f}/50)")

simulate_exam_pacing()
```

## 5. Hands-On Exercise: Practice Diagnostic Tracker

After completing each chapter's review questions in the book, enter your chapter quiz score below to compute your **Weighted Exam Readiness Score** across the 5 official domains.

```python
# Exercise: Update these chapter quiz scores (0.0 to 100.0) as you progress through the book!
chapter_quiz_scores = {
    "Ch 1": 90.0,
    "Ch 2": 80.0,
    "Ch 3": 85.0,
    "Ch 4": 75.0,
    "Ch 5": 88.0,
    "Ch 6": 92.0,
    "Ch 7": 82.0,
    "Ch 8": 78.0,
}

def compute_weighted_exam_readiness(scores: dict) -> float:
    domain_scores = {
        "Core Machine Learning & AI Knowledge": 0.5 * scores["Ch 2"] + 0.5 * scores["Ch 8"],
        "Software Development": 0.2 * scores["Ch 1"] + 0.8 * scores["Ch 3"],
        "Experimentation": 0.55 * scores["Ch 4"] + 0.45 * scores["Ch 7"],
        "Data Analysis & Visualization": scores["Ch 5"],
        "Trustworthy AI": scores["Ch 6"],
    }
    weights = {
        "Core Machine Learning & AI Knowledge": 0.30,
        "Software Development": 0.24,
        "Experimentation": 0.22,
        "Data Analysis & Visualization": 0.14,
        "Trustworthy AI": 0.10,
    }
    total_weighted = sum(domain_scores[d] * weights[d] for d in weights)
    print("Domain-Level Readiness Breakdown:")
    for d, w in weights.items():
        print(f"  - {d:<38} (Weight {int(w*100):>2}%): {domain_scores[d]:.1f}%")
    print("-" * 65)
    print(f"Overall Blueprint-Weighted Readiness Score: {total_weighted:.2f}%")
    print("Status:", "READY FOR EXAM (>= 80%)" if total_weighted >= 80.0 else "KEEP PRACTICING (< 80%)")
    return total_weighted

compute_weighted_exam_readiness(chapter_quiz_scores)
```
